```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

for i in df.columns:
    if i != "target":
        if df[i].dtype == 'int64' or df[i].dtype == 'float64':
            df[i] = scaler.fit_transform(df[i].values.reshape(-1,1))
df.head()
```

```python
from scipy.stats import iqr
q1=df['trestbps'].quantile(0.25)
q3=df['trestbps'].quantile(0.75)
IQR=q3-q1
lower_bound=q1-1.5*IQR
upper_bound=q3+1.5*IQR
df['trestbps']=np.where(df['trestbps']<lower_bound,lower_bound,df['trestbps'])
df['trestbps']=np.where(df['trestbps']>upper_bound,upper_bound,df['trestbps'])
print(f"'trestbps'column capped between {lower_bound} and {upper_bound},")
print("updated set")
print(df['trestbps'].describe())
```

```python
x=df.drop('target',axis=1)
y=df['target']
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_scaled = scaler.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x_scaled,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
df.head()
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
rf.fit(x_train,y_train)
```

```python
y_pred=rf.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score

print(
    accuracy_score(
        y_test,
        y_pred
    )
)
```

```python
from sklearn.metrics import confusion_matrix

print(
    confusion_matrix(
        y_test,
        y_pred
    )
)
```

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_test,
        y_pred
    )
)
```

```python
print(df.shape)
```

```python
print("Train Accuracy:", rf.score(x_train, y_train))
print("Test Accuracy:", rf.score(x_test, y_test))
```
